```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, Dataset
from torchvision import transforms
from transformers import CLIPProcessor, CLIPModel
from PIL import Image
import os
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score
from tqdm.notebook import tqdm
import numpy as np
import cv2
import segmentation_models_pytorch as smp
import albumentations as A
from albumentations.pytorch import ToTensorV2
from sklearn.utils.class_weight import compute_class_weight
```

```python
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {DEVICE}")

TRAIN_DIR_BASE = 'data/train'
TRAIN_CSV_PATH = 'data/train/train.csv'

TEST_IMAGE_DIR = 'data/test/images'
TEST_CSV_PATH = 'data/test/test.csv'

RESULT_DIR = 'data/result'
SUBMISSION_CSV_PATH = 'data/result/submission.csv'
SEGMENTATION_DIR = 'data/result/segmentation'
BEST_CLS_MODEL_PATH = 'data/best_classifier_model.pth'
BEST_SEG_MODEL_PATH = 'data/best_segmentation_model.pth'

os.makedirs(RESULT_DIR, exist_ok = True)
os.makedirs(SEGMENTATION_DIR, exist_ok = True)

CLASSES = ['class_benign', 'class_malignant', 'class_normal']
CLASS_MAP = {name: i for i, name in enumerate(CLASSES)}

CLASSES,CLASS_MAP
```

```python
def get_paths_and_label(row):
    img_name = row['image']
    base_name, ext = os.path.splitext(img_name)

    if row['class_benign'] == 1:
        label_name = 'benign'
        label_idx = CLASS_MAP['class_benign']

    elif row['class_malignant'] == 1:
        label_name = 'malignant'
        label_idx = CLASS_MAP['class_malignant']

    else:
        label_name = 'normal'
        label_idx = CLASS_MAP['class_normal']

    mask_name = f"{base_name}_mask{ext}"
    mask_path = os.path.join(TRAIN_DIR_BASE, label_name, 'masks', mask_name)
    img_path = os.path.join(TRAIN_DIR_BASE, label_name, 'images', img_name)
    return img_path, mask_path, label_idx

all_data_df = pd.read_csv(TRAIN_CSV_PATH)
paths_labels = all_data_df.apply(get_paths_and_label, axis=1, result_type='expand')
paths_labels.columns = ['img_path', 'mask_path', 'label_idx']
all_data_df = pd.concat([all_data_df, paths_labels], axis=1)

train_df, val_df = train_test_split(all_data_df, test_size = 0.2, random_state = 42, stratify = all_data_df['label_idx'])
train_df
```

```python
class XRayDataset(Dataset):
    def __init__(self, df, transform = None):
        self.df = df
        self.transform = transform

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]

        img = cv2.imread(row['img_path'])
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

        if pd.isna(row['mask_path']) or row['mask_path'] is None:
            mask = np.zeros(img.shape[:2], dtype = np.float32)
        else:
            mask = cv2.imread(row['mask_path'], cv2.IMREAD_GRAYSCALE)
            if mask is None:
                mask = np.zeros(img.shape[:2], dtype = np.float32)
            else:
                mask = mask.astype(np.float32) / 255.0

        label = torch.tensor(row['label_idx'], dtype = torch.long)

        if self.transform:
            augmented = self.transform(image = img, mask = mask)
            img = augmented['image']
            mask = augmented['mask']

        mask = mask.unsqueeze(0)
        return img, mask, label

IMG_HEIGHT , IMG_WIDTH = 224 , 224

train_transform = A.Compose(
    [
        # Resize the image to fixed dimensions (IMG_HEIGHT, IMG_WIDTH)
        A.Resize(height = IMG_HEIGHT, width = IMG_WIDTH),

        # Flip the image horizontally (left to right) with a 50% probability
        A.HorizontalFlip(p = 0.5),

        # Flip the image vertically (up to down) with a 50% probability
        A.VerticalFlip(p = 0.5),

        # Randomly rotate the image by 90, 180, or 270 degrees
        A.RandomRotate90(p = 0.5),

        # Apply random scaling, moving (translation), and rotation
        A.Affine(scale = (0.9, 1.1), translate_percent = (-0.0625, 0.0625), rotate = (-45, 45), p = 0.5),

        # Apply ONLY ONE of the transformations inside this block with a 30% probability
        A.OneOf([
            # 1. Apply elastic deformation (stretching like rubber)
            A.ElasticTransform(p = 0.5, alpha = 120, sigma = 120 * 0.05),

            # 2. Apply grid distortion (warping the image grid)
            A.GridDistortion(p=  0.5),
        ], p = 0.3),

        # Randomly change the brightness and contrast of the image
        A.RandomBrightnessContrast(brightness_limit=0.1, contrast_limit=0.1, p=0.3),

        # Standardize pixel values using ImageNet mean and standard deviation
        A.Normalize(
            mean = [0.485, 0.456, 0.406],
            std = [0.229, 0.224, 0.225],
            max_pixel_value = 255.0,
        ),

        # Convert the image array into a PyTorch Tensor
        ToTensorV2(),
    ]
)


val_transform = A.Compose([
    A.Resize(height = IMG_HEIGHT, width = IMG_WIDTH),
    A.Normalize(
        mean = [0.485, 0.456, 0.406],
        std = [0.229, 0.224, 0.225],
        max_pixel_value = 255.0,
    ),
    ToTensorV2()
])

train_dataset = XRayDataset(train_df, transform=  train_transform)
val_dataset = XRayDataset(val_df, transform = val_transform)

BATCH_SIZE = 16
train_loader = DataLoader(train_dataset, batch_size = BATCH_SIZE, shuffle = True)
val_loader = DataLoader(val_dataset, batch_size = BATCH_SIZE, shuffle = False)
```

```python
print("\n--- Starting Classification Training (CLIP VLM) ---")
import logging
    
import os
    
os.environ["HF_HUB_DISABLE_XET"] = "1"
os.environ["HF_HUB_DISABLE_PROGRESS_BARS"] = "1"

logging.basicConfig(level = logging.INFO)

MODEL_ID = "openai/clip-vit-base-patch32"
clip_model = CLIPModel.from_pretrained(MODEL_ID).to(DEVICE)
processor = CLIPProcessor.from_pretrained(MODEL_ID)
```

```python
text_prompts = [
    "an X-ray image of a benign tumor",
    "an X-ray image of a malignant tumor",
    "a normal X-ray image"
]
text_inputs = processor(text = text_prompts, return_tensors = "pt", padding = True, truncation = True).to(DEVICE)
```

```python
for param in clip_model.parameters():
    param.requires_grad = False

for param in clip_model.vision_model.encoder.layers[-1].parameters():
    param.requires_grad = True

if hasattr(clip_model, 'visual_projection'):
    for param in clip_model.visual_projection.parameters():
        param.requires_grad = True
```

```python
cls_optimizer = optim.AdamW(filter(lambda p: p.requires_grad, clip_model.parameters()), lr = 5e-5, weight_decay = 1e-4)

train_labels_list = train_df['label_idx'].values
class_weights = compute_class_weight(class_weight = 'balanced', classes = np.unique(train_labels_list), y = train_labels_list)
class_weights_tensor = torch.tensor(class_weights, dtype = torch.float32).to(DEVICE)

cls_criterion = nn.CrossEntropyLoss(weight = class_weights_tensor)
```

```python
torch.cuda.empty_cache()
```

```python
best_f1 = 0.0
EPOCHS_CLS = 30
cls_scheduler = optim.lr_scheduler.ReduceLROnPlateau(cls_optimizer, mode = 'max', factor = 0.5, patience = 3)
cls_scaler = torch.amp.GradScaler('cuda')

for epoch in range(EPOCHS_CLS):
    clip_model.train()
    train_loss = 0.0
    pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{EPOCHS_CLS} [CLS Train]")

    for images, masks, labels in pbar:
        images, labels = images.to(DEVICE), labels.to(DEVICE)

        cls_optimizer.zero_grad()

        with torch.amp.autocast('cuda'):
            outputs = clip_model(
                pixel_values = images,
                input_ids = text_inputs.input_ids,
                attention_mask = text_inputs.attention_mask
            )
            logits = outputs.logits_per_image
            loss = cls_criterion(logits, labels)

        cls_scaler.scale(loss).backward()
        cls_scaler.step(cls_optimizer)
        cls_scaler.update()

        train_loss += loss.item()
        pbar.set_postfix({"Loss": loss.item()})

    clip_model.eval()
    all_preds, all_labels = [], []
    val_loss = 0.0

    with torch.no_grad():
        for images, masks, labels in val_loader:
            images, labels = images.to(DEVICE), labels.to(DEVICE)

            with torch.amp.autocast('cuda'):
                outputs = clip_model(
                    pixel_values = images,
                    input_ids = text_inputs.input_ids,
                    attention_mask = text_inputs.attention_mask
                )
                logits = outputs.logits_per_image
                loss = cls_criterion(logits, labels)

            val_loss += loss.item()
            preds = torch.argmax(logits, dim=1)
            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())

    f1 = f1_score(all_labels, all_preds, average='macro')
    avg_val_loss = val_loss / len(val_loader)

    print(f"Val Loss: {avg_val_loss:.4f} | Val F1 Macro: {f1:.4f}")

    cls_scheduler.step(f1)

    if f1 > best_f1:
        best_f1 = f1
        torch.save(clip_model.state_dict(), BEST_CLS_MODEL_PATH)
        print("--> Best CLS model saved!")
```

```python
best_val_iou = 0.0
EPOCHS_SEG = 50

seg_model = smp.Unet(
    encoder_name="efficientnet-b4",
    encoder_weights="imagenet",
    in_channels=3,
    classes=1,
).to(DEVICE)

seg_optimizer = optim.Adam(seg_model.parameters(), lr = 1e-4, weight_decay = 1e-5)
scheduler = optim.lr_scheduler.ReduceLROnPlateau(seg_optimizer, mode = 'max', factor = 0.5, patience = 3)

bce_loss = nn.BCEWithLogitsLoss()
dice_loss = smp.losses.DiceLoss(mode='binary', eps=1e-4)

scaler = torch.amp.GradScaler('cuda')
```

```python
def calculate_iou(preds, masks, threshold=0.5):
    preds = torch.sigmoid(preds) > threshold
    masks = masks > 0.5

    intersection = (preds & masks).sum(dim = (1, 2, 3)).float()
    union = (preds | masks).sum(dim=(1, 2, 3)).float()

    iou = torch.where(union == 0, torch.tensor(1.0, device = preds.device), intersection / union)

    return iou.mean().item()
```

```python
torch.cuda.empty_cache()
```

```python
print("\n--- Starting Segmentation Training (U-Net) ---")
for epoch in range(EPOCHS_SEG):
    seg_model.train()
    train_loss = 0.0

    pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{EPOCHS_SEG} [SEG Train]")
    for images, masks, labels in pbar:
        images, masks = images.to(DEVICE), masks.to(DEVICE)

        seg_optimizer.zero_grad()

        with torch.amp.autocast('cuda'):
            outputs = seg_model(images)

        outputs_fp32 = outputs.float()
        masks_fp32 = masks.float()
        loss = bce_loss(outputs_fp32, masks_fp32) + dice_loss(outputs_fp32, masks_fp32)

        scaler.scale(loss).backward()
        scaler.unscale_(seg_optimizer)
        torch.nn.utils.clip_grad_norm_(seg_model.parameters(), max_norm=1.0)
        scaler.step(seg_optimizer)
        scaler.update()

        train_loss += loss.item()
        pbar.set_postfix({"Loss": loss.item()})

    seg_model.eval()
    val_loss = 0.0
    val_iou = 0.0

    with torch.no_grad():
        for images, masks, labels in val_loader:
            images, masks = images.to(DEVICE), masks.to(DEVICE)

            with torch.amp.autocast('cuda'):
                outputs = seg_model(images)

            outputs_fp32 = outputs.float()
            masks_fp32 = masks.float()
            loss = bce_loss(outputs_fp32, masks_fp32) + dice_loss(outputs_fp32, masks_fp32)

            val_loss += loss.item()
            val_iou += calculate_iou(outputs, masks)

    val_loss /= len(val_loader)
    val_iou /= len(val_loader)

    print(f"Val Loss: {val_loss:.4f} | Val IoU: {val_iou:.4f}")

    scheduler.step(val_iou)

    if val_iou > best_val_iou:
        best_val_iou = val_iou
        torch.save(seg_model.state_dict(), BEST_SEG_MODEL_PATH)
        print("--> Best model saved!")
```

```python
print("\n--- Running Inference on Test Set ---")

class XRayTestDataset(Dataset):
    def __init__(self, df, img_dir, transform=None):
        self.df = df
        self.img_dir = img_dir
        self.transform = transform

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        img_name = self.df.iloc[idx]['image']
        img_path = os.path.join(self.img_dir, img_name)
        img = cv2.imread(img_path)

        orig_shape = img.shape[:2]
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

        if self.transform:
            img = self.transform(image=img)['image']

        return img, img_name, orig_shape

test_df = pd.read_csv(TEST_CSV_PATH)
test_dataset = XRayTestDataset(test_df, TEST_IMAGE_DIR, transform=val_transform)
test_loader = DataLoader(test_dataset, batch_size=1, shuffle=False)

clip_model.load_state_dict(torch.load(BEST_CLS_MODEL_PATH))
clip_model.eval()

seg_model.load_state_dict(torch.load(BEST_SEG_MODEL_PATH))
seg_model.eval()

results = []

with torch.no_grad():
    for img, img_name, orig_shape in tqdm(test_loader, desc="Testing"):
        img = img.to(DEVICE)
        img_name = img_name[0]
        h, w = orig_shape[0].item(), orig_shape[1].item()

        outputs_cls = clip_model(
            pixel_values=img,
            input_ids=text_inputs.input_ids,
            attention_mask=text_inputs.attention_mask
        )
        pred_idx = torch.argmax(outputs_cls.logits_per_image, dim=1).item()
        results.append({'image': img_name, 'prediction_idx': pred_idx})

        outputs_seg = seg_model(img)
        pred_mask = torch.sigmoid(outputs_seg)
        pred_mask = (pred_mask > 0.5).float()
        mask_np = pred_mask.squeeze().cpu().numpy() * 255.0
        mask_np = mask_np.astype(np.uint8)

        mask_np = cv2.resize(mask_np, (w, h), interpolation=cv2.INTER_NEAREST)

        base_name, ext = os.path.splitext(img_name)
        save_path = os.path.join(SEGMENTATION_DIR, f"{base_name}_mask{ext}")
        cv2.imwrite(save_path, mask_np)
```

```python
results_df = pd.DataFrame(results)
results_df['class_benign'] = (results_df['prediction_idx'] == CLASS_MAP['class_benign']).astype(int)
results_df['class_malignant'] = (results_df['prediction_idx'] == CLASS_MAP['class_malignant']).astype(int)
results_df['class_normal'] = (results_df['prediction_idx'] == CLASS_MAP['class_normal']).astype(int)

submission = pd.merge(test_df[['image']], results_df, on='image', how='left')
submission = submission[['image', 'class_benign', 'class_malignant', 'class_normal']]
submission.to_csv(SUBMISSION_CSV_PATH, index=False)

print(f"\n✅ All tasks completed.")
print(f"Submission saved to: {SUBMISSION_CSV_PATH}")
print(f"Masks saved to: {SEGMENTATION_DIR}")
```

```python
import cv2
import os
import pandas as pd
import numpy as np

folder_path = '/content/segmentation'
reference_folder = '/content/test/images'
csv_path = '/content/submission.csv'

df = pd.read_csv(csv_path)
normal_status = dict(zip(df['image'], df['class_normal']))

valid_extensions = ('.png', '.jpg', '.jpeg', '.bmp', '.tif', '.tiff')

for filename in os.listdir(folder_path):
    if filename.lower().endswith(valid_extensions):
        file_path = os.path.join(folder_path, filename)

        ref_filename = filename.replace('_mask', '')
        ref_path = os.path.join(reference_folder, ref_filename)

        ref_img = cv2.imread(ref_path)

        if ref_img is None:
            print(f"Failed to read reference image: {ref_filename}")
            continue

        ref_height, ref_width = ref_img.shape[:2]

        img = cv2.imread(file_path)

        if img is not None:
            resized_img = cv2.resize(img, (ref_width, ref_height))

            if normal_status.get(ref_filename, 0) == 1:
                resized_img = np.zeros_like(resized_img)
                status_msg = " (Blacked out - Normal Class)"
            else:
                status_msg = ""

            cv2.imwrite(file_path, resized_img)
            print(f"Processed {filename}: Resized to {ref_width}x{ref_height}{status_msg}")
        else:
            print(f"Failed to read mask: {filename}")

print("تمام تصاویر با موفقیت پردازش شدند.")
```
